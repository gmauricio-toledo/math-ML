El dataset contiene individuos de ambos sexos.
Para simplificar, nos enfocamos en un solo grupo.

Extraemos las dos variables que nos interesan: fémur izquierdo y altura.
Eliminamos filas incompletas (valores nulos o ceros que no tienen sentido físico).
Convertimos a array de NumPy: cada fila es un punto en el plano,
la primera coordenada es el fémur, la segunda es la altura.

```python
import numpy as np

gender = 'Male'

df_aux = df[df['Sex']==gender].copy()
df_data = df_aux[['Femur left','Height in grave']].copy()
df_data.dropna(inplace=True) # Eliminar valores faltantes
df_data = df_data[df_data['Femur left']!=0].copy()
df_data = df_data[df_data['Height in grave']!=0].copy()

datos = df_data.values
datos.shape
```

Estos 30-50 puntos son nuestros datos. Entrenar un modelo significa
encontrar una función (por ejemplo, una línea recta) que pase
"lo más cerca posible" de todos estos puntos.
El objetivo es aprender una relación que sirva para predecir altura
a partir de fémur en individuos que no vimos durante el entrenamiento.

```python
datos
```

```python
import matplotlib.pyplot as plt

plt.figure(dpi=150)
plt.suptitle("Relación entre el fémur y la altura")
plt.scatter(datos[:,0],datos[:,1],color='red')
plt.xlabel('Largo del fémur (cm)',fontsize=14)
plt.ylabel('Altura (cm)',fontsize=14)
plt.show()
```

El mejor modelo de la familia

$$\{ f_{a,b}(x):=ax+b\;|\; a,b\in\mathbb{R}\}$$

está dado por los parámetros:

$$a = \frac{(\sum_{i=1}^{n} x_i y_i) - n\bar{x}\bar{y}}{(\sum_{i=1}^{n} x_i^2) - n\bar{x}^2}$$

$$b = \bar{y} - \beta_1\bar{x}$$

donde $\bar{x}$ es el promedio de los valores $x_i$.
