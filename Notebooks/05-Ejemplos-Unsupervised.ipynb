# Aprendizaje no supervisado

Dado un conjunto de datos $X\subset\mathbb{R}^N$, el entrenamiento en el aprendizaje no supervisado consiste en encontrar un modelo

$$f_0\in \mathcal{F}=\{f:\mathbb{R}^N\rightarrow E\}$$

que minimice una función de costo intrínseca (que no depende de etiquetas externas):

$$\mathcal{L}:\mathcal{F}\times \mathcal{P}(X) \rightarrow \mathbb{R}$$

donde $E$ es el espacio de salida que depende de la tarea:
- **Reducción de dimensionalidad**: $E=\mathbb{R}^M$ con $M \leq N$
- **Clustering**: $E = \{1,2,\ldots,K\}$ (etiquetas de clusters)

Formalmente:
$$f_0 = \arg\min_{f\in\mathcal{F}} \mathcal{L}(f, X)$$

## Ejemplo: Reducción de dimensionalidad


```python
import numpy as np
import matplotlib.pyplot as plt

num_points = 1000

xs = np.random.standard_normal(size=(num_points,))
ys_1 = 2*xs - 3
ys_2 = 2*xs - 3 + np.random.normal(loc=0, scale=0.25, size=(num_points,))
ys_3 = 0.5*xs**2 - xs  + 1 + np.random.normal(loc=0, scale=0.2, size=(num_points,))
ys_4 = np.zeros_like(xs) # np.random.normal(loc=0, scale=0.15, size=(num_points,))


plt.figure(figsize=(10, 10))
plt.subplot(2, 2, 1)
plt.scatter(xs, ys_4, c=xs)
plt.subplot(2, 2, 2)
plt.scatter(xs, ys_1, c=xs)
plt.subplot(2, 2, 3)
plt.scatter(xs, ys_2, c=xs)
plt.subplot(2, 2, 4)
plt.scatter(xs, ys_3, c=xs)
plt.axis('equal')
plt.show()
```

## Ejemplo: Clustering

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np

X, colors = make_blobs(random_state=3,
                  centers=3,
                  n_samples=100,
                  n_features=2,
                  )

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1],c='black')
plt.title('Los datos')

plt.subplot(1, 2, 2)
for i in np.unique(colors):
    plt.scatter(X[colors == i, 0], X[colors == i, 1],
                label=f'Cluster {i}')
plt.title('Los datos pasados por el modelo')
plt.legend(loc='best')
plt.show()
```

## Ejemplo: Clustering + Reducción de dimensionalidad




```python
import pandas as pd

url = "https://raw.githubusercontent.com/DCDPUAEM/DCDP/main/03%20Machine%20Learning/data/Mall_Customers.csv"

mall_df = pd.read_csv(url)
mall_df.drop(columns=['CustomerID'],inplace=True)
original_mall_df = mall_df.copy()
mall_df
```

🔵 ¿Qué hacemos con la variable `Gender`?

```python
mall_df = pd.get_dummies(mall_df,drop_first=True, dtype=int)
mall_df
```

```python
import numpy as np

X = mall_df.values
X.shape
```

```python
mall_df.describe()
```

```python
import seaborn as sns

sns.pairplot(mall_df)
```

Hacemos clustering:

```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans

X = MinMaxScaler().fit_transform(X)

clustering = KMeans(n_clusters=4)
clustering.fit(X)
clusters = clustering.labels_
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
clusetrs = clustering.labels_

plt.figure()
for i in np.unique(clusters):
    plt.scatter(X_pca[clusters == i, 0], X_pca[clusters == i, 1],
                label=f'Cluster {i}')
plt.legend(loc='best')
plt.show()
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(n_components=2)
X_tsne = tsne.fit_transform(X)

plt.figure()
for i in np.unique(clusters):
    plt.scatter(X_tsne[clusters == i, 0], X_tsne[clusters == i, 1],
                label=f'Cluster {i}')
plt.legend(loc='best')
plt.show()
```

Interpretemos estos clusters:

```python
from ipywidgets import interact, IntSlider, Dropdown
import ipywidgets as widgets

num_clusters = np.unique(clusters).shape[0]

@interact(cluster=Dropdown(
    options=list(range(num_clusters)),
    description='Cluster:'
))

def show_cluster(cluster):
    display(original_mall_df[clusters==cluster])
```
