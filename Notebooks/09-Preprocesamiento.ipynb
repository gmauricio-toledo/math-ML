# Clases

Classes provide a means of bundling data and functionality together. *Una clase es un conjunto de variables y funciones agrupadas que interactuan entre ellas dentro de la clase*

Creating a new class creates a new type of object, allowing new instances of that type to be made.

Each class instance can have attributes attached to it for maintaining its state. Class instances can also have methods (defined by its class) for modifying its state.

```python
import numpy as np

class Estudiante:
    def __init__(self, nombre, matricula, calificaciones):
        self.nombre = nombre
        self.matricula = matricula
        self.calificaciones = calificaciones

    def promedio(self):
        return sum(self.calificaciones) / len(self.calificaciones)

    def materias_aprobadas(self, calificacion_minima):
        aprobadas = np.where(np.array(self.calificaciones) >= calificacion_minima, 1, 0)
        return sum(aprobadas)
```

# Imputación


La imputación es el proceso de reemplazar los valores faltantes en un conjunto de datos con valores sustitutos. Esto es crucial porque muchos algoritmos de aprendizaje automático no pueden manejar valores `NaN` (Not a Number) y eliminar filas o columnas enteras con valores faltantes podría resultar en la pérdida de información valiosa.

### Estrategias Comunes de Imputación:

1.  **Media/Mediana:** Para datos numéricos, los valores faltantes se pueden reemplazar con la media o la mediana de la columna.
2.  **Moda:** Para datos categóricos, los valores faltantes se pueden reemplazar con el valor más frecuente (moda) de la columna.
3.  **Valor Constante/Categoría 'Missing':** Especialmente útil para columnas con muchos valores faltantes, donde se puede crear una nueva categoría 'Missing' o usar un valor constante específico.
4.  **Imputación Predictiva:** Usar un modelo de aprendizaje automático para predecir los valores faltantes basándose en otras columnas.

```python
import pandas as pd

titanic_df = pd.read_csv('Titanic-Dataset.csv')
titanic_df
```

### Identificación de Valores Faltantes

Primero, veamos nuevamente las columnas con valores faltantes y sus conteos para entender dónde necesitamos aplicar la imputación.

```python
print('Valores faltantes antes de la imputación:')
display(titanic_df.isnull().sum())
```

Observamos que las columnas 'Age' (Edad), 'Cabin' (Cabina) y 'Embarked' (Embarque) tienen valores faltantes.

### Imputación de la columna 'Age' (Edad)

'Age' es una columna numérica. Podemos imputar sus valores faltantes usando la **media** o la **mediana**. La mediana es a menudo más robusta a los valores atípicos (outliers) que la media. Usaremos la mediana en este caso.

```python
print('Valores faltantes en Age antes de imputación:', titanic_df['Age'].isnull().sum())

# Calculamos la mediana de la columna 'Age'
median_age = titanic_df['Age'].median()

# Imputamos los valores faltantes con la mediana
titanic_df['Age'] = titanic_df['Age'].fillna(median_age)

print('Valores faltantes en Age después de imputación:', titanic_df['Age'].isnull().sum())
```

### Imputación de la columna 'Embarked' (Embarque)

'Embarked' es una columna categórica. La mejor estrategia es imputar los valores faltantes con la **moda** (el valor más frecuente) de la columna.

```python
print('Valores faltantes en Embarked antes de imputación:', titanic_df['Embarked'].isnull().sum())

# Calculamos la moda de la columna 'Embarked'
mode_embarked = titanic_df['Embarked'].mode()[0] # .mode() devuelve una Serie, tomamos el primer valor

# Imputamos los valores faltantes con la moda
titanic_df['Embarked'] = titanic_df['Embarked'].fillna(mode_embarked)

print('Valores faltantes en Embarked después de imputación:', titanic_df['Embarked'].isnull().sum())
```

### Imputación de la columna 'Cabin' (Cabina)

La columna 'Cabin' tiene un gran número de valores faltantes. En estos casos, imputar con la moda o la media/mediana no sería apropiado. Una estrategia común es crear una nueva categoría 'Missing' (Faltante) o 'Unknown' (Desconocido) para estos valores, lo que permite que el modelo los trate como una categoría separada.

```python
print('Valores faltantes en Cabin antes de imputación:', titanic_df['Cabin'].isnull().sum())

# Imputamos los valores faltantes con la categoría 'Missing'
titanic_df['Cabin'] = titanic_df['Cabin'].fillna('Missing')

print('Valores faltantes en Cabin después de imputación:', titanic_df['Cabin'].isnull().sum())
```

### Verificación Final de Valores Faltantes

Después de aplicar las diferentes estrategias de imputación, verificamos que no queden valores faltantes en el DataFrame.

```python
print('Valores faltantes después de la imputación completa:')
display(titanic_df.isnull().sum())
```

Como puedes ver, todas las columnas han sido imputadas y ya no hay valores faltantes en el DataFrame.

## One-Hot Encoding

One-Hot Encoding es un proceso para convertir variables categóricas nominales en un formato que puede ser entendido por algoritmos de machine learning. En lugar de tener una sola columna con múltiples categorías, se crea una nueva columna binaria (0 o 1) para cada categoría única. Esto evita que el modelo interprete erróneamente un orden o jerarquía entre las categorías que no existe.

```python
# Aplicamos One-Hot Encoding a la columna 'Sex' y 'Embarked'
titanic_df_encoded = pd.get_dummies(titanic_df,
                                    columns=['Sex', 'Embarked'],
                                    drop_first=True)

display(titanic_df_encoded.head())
```

Al aplicar One-Hot Encoding, si mantenemos una columna binaria para cada categoría de una variable original, podemos introducir un problema conocido como **multicolinealidad**. Esto ocurre cuando una variable puede ser perfectamente predicha por una combinación lineal de otras variables en el modelo.

Por ejemplo, para la columna 'Sex' con categorías 'male' y 'female':

*   Si creamos 'Sex_male' y 'Sex_female':
    *   Cuando 'Sex_male' es 1, 'Sex_female' es 0.
    *   Cuando 'Sex_male' es 0, 'Sex_female' es 1.

Esto significa que la columna 'Sex_female' es simplemente
$$\text{Sex_female} = 1 - \text{Sex_male}$$

Ambas columnas están perfectamente correlacionadas negativamente, y una es redundante si se incluyen ambas en un modelo lineal (como la regresión lineal o logística).

Para evitar esto, se utiliza el parámetro `drop_first=True` en `pd.get_dummies()`. Este argumento elimina la primera columna binaria creada para cada variable categórica. De esta manera, si tenemos `k` categorías, solo creamos `k-1` columnas binarias. La información de la categoría eliminada se representa implícitamente cuando todas las demás columnas binarias son cero.

En nuestro caso:
*   Para 'Sex', se crea `Sex_male`. La categoría 'female' se infiere cuando `Sex_male` es 0.
*   Para 'Embarked' (C, Q, S), se crean `Embarked_Q` y `Embarked_S`. La categoría 'C' se infiere cuando `Embarked_Q` es 0 y `Embarked_S` es 0.

Esto ayuda a prevenir la multicolinealidad perfecta y a mejorar la estabilidad y la interpretabilidad de algunos modelos de aprendizaje automático.
