# 1. El dataset y EDA

```python
from numpy import test
import pandas as pd

train_path = '/content/sample_data/mnist_train_small.csv'
test_path = '/content/sample_data/mnist_test.csv'

train_df = pd.read_csv(train_path, header=None)
X_train = train_df.iloc[:, 1:].values.astype('float32')
y_train = train_df[0].values.astype('float32')

test_df = pd.read_csv(test_path, header=None)
X_test = test_df.iloc[:, 1:].values
y_test = test_df[0].values

print(f'X_train: {X_train.shape}')
print(f'X_test: {X_test.shape}')
print(f'y_train: {y_train.shape}')
print(f'y_test: {y_test.shape}')
```

```python
train_df.describe()
```

```python
import matplotlib.pyplot as plt
import numpy as np

random_idxs = np.random.choice(X_train.shape[0],4)

fig, axs = plt.subplots(2, 2, figsize=(6, 6))
for i, idx in enumerate(random_idxs):
    axs[i // 2, i % 2].imshow(X_train[idx].reshape(28, 28), cmap='gray')
    axs[i // 2, i % 2].set_title(f'Label: {y_train[idx]}')
    axs[i // 2, i % 2].axis('off')

plt.tight_layout()
plt.show()


print("Las features:")
for idx in random_idxs:
    print(X_train[idx,:20])
```

```python
import collections

train_countings = collections.Counter(y_train)
print(train_countings)

test_countings = collections.Counter(y_test)
```

# 3. El Preprocesamiento

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

🔵 ¿Por qué no usamos `StandardScaler` o `MinMaxScaler`?

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scl = scaler.fit_transform(X_train)

random_idxs = np.random.choice(X_train_scl.shape[0],4)

fig, axs = plt.subplots(2, 2, figsize=(6, 6))
fig.suptitle('Datos escalados con StandardScaler')
for i, idx in enumerate(random_idxs):
    axs[i // 2, i % 2].imshow(X_train_scl[idx].reshape(28, 28), cmap='gray')
    axs[i // 2, i % 2].set_title(f'Label: {y_train[idx]}')
    axs[i // 2, i % 2].axis('off')

plt.tight_layout()
plt.show()

fig, axs = plt.subplots(2, 2, figsize=(6, 6))
fig.suptitle('Datos originales')
for i, idx in enumerate(random_idxs):
    axs[i // 2, i % 2].imshow(X_train[idx].reshape(28, 28), cmap='gray')
    axs[i // 2, i % 2].set_title(f'Label: {y_train[idx]}')
    axs[i // 2, i % 2].axis('off')

plt.tight_layout()
plt.show()
```

# 4. Entrenamiento

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

y_train_pred = knn.predict(X_train)
y_test_pred = knn.predict(X_test)
```

# 5. Evaluación

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score

print(f'Train accuracy: {accuracy_score(y_train, y_train_pred)}')
print(f'Test accuracy: {accuracy_score(y_test, y_test_pred)}')
print(60*"-")
print(f'Train recall: {recall_score(y_train, y_train_pred, average="macro")}')
print(f'Test recall: {recall_score(y_test, y_test_pred, average="macro")}')
print(60*"-")
print(f'Train precision: {precision_score(y_train, y_train_pred, average="macro")}')
print(f'Test precision: {precision_score(y_test, y_test_pred, average="macro")}')
print(60*"-")
print(f'Train f1: {f1_score(y_train, y_train_pred, average="macro")}')
print(f'Test f1: {f1_score(y_test, y_test_pred, average="macro")}')
```

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns

cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=range(10), yticklabels=range(10))
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_test_pred))
```
