## Información del Conjunto de Datos PIMA Indians Diabetes

Este conjunto de datos se originó en el National Institute of Diabetes and Digestive and Kidney Diseases. El objetivo es predecir si un paciente tiene diabetes o no basándose en ciertas mediciones de diagnóstico incluidas en el conjunto de datos. Todas las pacientes aquí son mujeres de al menos 21 años de edad de ascendencia Pima.

### Significado de cada columna:

*   **Pregnancies**: Número de veces embarazada.
*   **Glucose**: Concentración de glucosa en plasma a las 2 horas en una prueba oral de tolerancia a la glucosa.
*   **BloodPressure**: Presión arterial diastólica (mm Hg).
*   **SkinThickness**: Grosor del pliegue cutáneo del tríceps (mm).
*   **Insulin**: Insulina sérica de 2 horas (mu U/ml).
*   **BMI**: Índice de masa corporal (peso en kg/(altura en m)^2).
*   **DiabetesPedigreeFunction**: Función del pedigrí de la diabetes (una función que cuantifica la probabilidad de diabetes basándose en antecedentes familiares).
*   **Age**: Edad (años).
*   **Outcome**: Variable de clase (0 si no tiene diabetes, 1 si tiene diabetes).
